```python
# Greedy Best-Fit Algorithm

def best_fit(items, bin_capacity):
    bins = []

    for item in items:
        best_bin = -1
        min_remaining = bin_capacity + 1

        # Find the bin with the least remaining space
        # after placing the item
        for i in range(len(bins)):
            if bins[i] + item <= bin_capacity:
                remaining = bin_capacity - (bins[i] + item)

                if remaining < min_remaining:
                    min_remaining = remaining
                    best_bin = i

        # If a suitable bin is found
        if best_bin != -1:
            bins[best_bin] += item
        else:
            # Create a new bin
            bins.append(item)

    return bins


# Input
items = [4, 8, 1, 4, 2, 1]
bin_capacity = 10

# Apply Best-Fit
bins = best_fit(items, bin_capacity)

# Output
print("Items:", items)
print("Bin Capacity:", bin_capacity)
print("Number of bins used:", len(bins))
print("Bins:", bins)
```
